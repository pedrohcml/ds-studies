```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
base = pd.read_csv('trees.csv')
base.head()
```

```python
plt.scatter(base.Girth, base.Volume)
```

```python
plt.scatter(base.Girth, base.Height)
```

```python
plt.scatter(base.Height, base.Volume, marker = '*')
```

```python
plt.hist(base.Volume)
```

```python
plt.figure(1)
plt.subplot(3,2,1)
plt.scatter(base.Girth, base.Volume)
plt.title('Circunferencia e Volume')
plt.subplot(3,2,2)
plt.scatter(base.Girth, base.Height)
plt.title('Circunferencia e Altura')
plt.subplot(3,2,5)
plt.scatter(base.Height, base.Volume, marker = '*')
plt.title('Altura e Volume')
plt.subplot(3,2,6)
plt.hist(base.Volume)
plt.title('Volume')
```
