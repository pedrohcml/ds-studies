```python
from groq import Groq
```

```python
client = Groq(api_key = 'randomkey')
model = 'openai/gpt-oss-20b'
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[
        {'role': 'user', 'content': 'Tell me a joke'}
    ]
)
```

```python
print(response)
```

```python
message = (response.choices[0].message.content)
print(message)
```
