```python
import matplotlib.pyplot as plt
import nltk
nltk.download('stopwords')
from nltk.corpus import PlaintextCorpusReader
from nltk.corpus import stopwords
from matplotlib.colors import ListedColormap
from wordcloud import WordCloud
import string
```

```python
#criando corpus. Parametros -> nome da pasta ou arquivo, buscar todos os tipos 
#de arquivos, codificacao da importacao dos arquivos
corpus = PlaintextCorpusReader('Arquivos', '.*', encoding = 'ISO-8859-1')
```

```python
#leitura dos arquivos
arquivos = corpus.fileids()
arquivos[0]
```

```python
#acesso a todas as palavras de todos os arquivos do corpus
todo_texto = corpus.raw()
todo_texto
```

```python
#obtencao de todas as palavras do corpus e visualizacao da quantidade
palavras = corpus.words()
#acessando pelo indice
palavras[180]
```

```python
#quantidade de palavras no corpus antes do pre-processamento
len(palavras)
```

```python
#obtancao das stopwords em ingles com o NLTK
stops = stopwords.words('english')
```

```python
mapa_cores = ListedColormap(['orange', 'green', 'red', 'magenta'])
nuvem = WordCloud(background_color = 'white',
                 colormap = mapa_cores,
                 stopwords = stops,
                 max_words = 100)

nuvem.generate(todo_texto)
plt.imshow(nuvem)
```

```python
#cria nova lista de palavras, removendo stop words
palavras_semstop = [p for p in palavras if p not in stops]
len(palavras_semstop)
```

```python
#remove pontuacoes, gerando uma lista sem stop words e sem pontuacao
palavras_sem_pontuacao = [p for p in palavras_semstop if p not in string.punctuation]
len(palavras_sem_pontuacao)
```

```python
frequencia = nltk.FreqDist(palavras_sem_pontuacao)
frequencia
```

```python
mais_comuns = frequencia.most_common(100)
mais_comuns
```
