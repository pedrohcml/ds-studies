```python
import psycopg2
```

```python
conexao = psycopg2.connect(host='localhost', database='AulaSQL', user='postgres', password='password', port=5432)
```

```python
#cria cursor
cursor = conexao.cursor()
#define consulta
consulta = 'select * from clientes'
#executa
cursor.execute(consulta)
```

```python
#recupera todos os registros
registros = cursor.fetchall()
```

```python
for row in registros:
    print(row[0])
    print('Nome - ', row[1]),
    print('Estado - ', row[2]),
    print('Status - ', row[3], '\n')
```

```python
#fecha conexao e cursor
cursor.close()
conexao.close()
```
