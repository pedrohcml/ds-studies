```python
from scipy.stats import norm
```

```python
#cdf(menor que, media, desvio padrao)
norm.cdf(6, 8, 2)
```

```python
#sf(maior que, media, desvio padrao)
norm.sf(6, 8, 2)
1 - norm.cdf(6, 8, 2)
```

```python
norm.cdf(6, 8, 2) + norm.sf(10, 8, 2)
```

```python
norm.cdf(10, 8, 2) - norm.sf(8, 8, 2)
```
