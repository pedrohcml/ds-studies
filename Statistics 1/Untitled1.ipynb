```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from xgboost import XGBClassifier

# 1. Gerando os dados simulados
np.random.seed(43)
n_clientes = 1000

data = pd.DataFrame({
    'idade': np.random.randint(18, 70, n_clientes),
    'score_credito': np.random.randint(300, 850, n_clientes),
    'meses_contrato': np.random.randint(1, 60, n_clientes)
})

# Regra com ruído para gerar o Churn
lin_comb = -0.02 * data['score_credito'] - 0.05 * data['meses_contrato'] + 12
probabilidade = 1 / (1 + np.exp(-lin_comb))
y = np.random.binomial(1, probabilidade)

# 2. Separando variáveis explicativas (X) do alvo (y)
X = data

# 3. Divisão de treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=43)

# 4. Rodando o XGBoost estruturado
xgb_model = XGBClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
xgb_model.fit(X_train, y_train)
xgb_predictions = xgb_model.predict(X_test)

print("Acurácia Real do XGBoost:", accuracy_score(y_test, xgb_predictions))
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier

# 1. Gerando a base de dados com nomes de variáveis exclusivos para o Random Forest
gerador_aleatorio = np.random.RandomState(42)
total_registros = 1000

tabela_clientes_rf = pd.DataFrame({
    'idade_consumidor': gerador_aleatorio.randint(18, 70, total_registros),
    'pontuacao_credito': gerador_aleatorio.randint(300, 850, total_registros),
    'tempo_fidelidade_meses': gerador_aleatorio.randint(1, 60, total_registros)
})

# Calculando a lógica do cancelamento
equacao_linear = -0.02 * tabela_clientes_rf['pontuacao_credito'] - 0.05 * tabela_clientes_rf['tempo_fidelidade_meses'] + 12
probabilidade_cancelar = 1 / (1 + np.exp(-equacao_linear))
alvo_churn_rf = gerador_aleatorio.binomial(1, probabilidade_cancelar)

# 2. Separando em preditores e alvo com nomenclatura nova
atributos_entrada_rf = tabela_clientes_rf
resposta_saida_rf = alvo_churn_rf

# 3. Divisão de treino e teste exclusiva
treino_x_rf, teste_x_rf, treino_y_rf, teste_y_rf = train_test_split(
    atributos_entrada_rf, 
    resposta_saida_rf, 
    test_size=0.2, 
    random_state=42
)

# 4. Inicializando e treinando o classificador Random Forest
modelo_floresta_aleatoria = RandomForestClassifier(n_estimators=100, random_state=42)
modelo_floresta_aleatoria.fit(treino_x_rf, treino_y_rf)

# 5. Calculando as previsões e a acurácia
predicoes_finais_rf = modelo_floresta_aleatoria.predict(teste_x_rf)
resultado_acuracia_rf = accuracy_score(teste_y_rf, predicoes_finais_rf)

print(f"Acurácia Real do Random Forest: {resultado_acuracia_rf:.3f}")

# 6. Extraindo a importância das variáveis para a floresta
pesos_atributos_rf = pd.DataFrame({
    'Atributo': atributos_entrada_rf.columns,
    'Peso_Importancia': modelo_floresta_aleatoria.feature_importances_
}).sort_values(by='Peso_Importancia', ascending=False)

print("\nImportância das Variáveis (Random Forest):")
print(pesos_atributos_rf)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report

# 1. Calculando cada métrica individualmente
acuracia = accuracy_score(teste_y_rf, predicoes_finais_rf)
precisao = precision_score(teste_y_rf, predicoes_finais_rf)
recall = recall_score(teste_y_rf, predicoes_finais_rf)

print("--- Métricas Individuais ---")
print(f"Acurácia: {acuracia:.3f}")
print(f"Precisão: {precisao:.3f}")
print(f"Recall:   {recall:.3f}")
print("\n" + "="*40 + "\n")

# 2. O Jeito Avançado: Relatório Completo de Classificação
# Essa função traz um resumo executivo completo de todas as métricas para cada classe (0 e 1)
print("--- Relatório Completo (Classification Report) ---")
print(classification_report(teste_y_rf, predicoes_finais_rf, target_names=['Ficou (0)', 'Cancelou (1)']))
```
