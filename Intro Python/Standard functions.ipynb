```python
from statistics import *
print("Media: ", mean(lst))
print("Mediana: ", median(lst))
print("Moda: ", mode(lst))
```

```python
print("Desvio padrao: ", stdev(lst))
```

```python
print("Variancia: ", variance(lst))
```
