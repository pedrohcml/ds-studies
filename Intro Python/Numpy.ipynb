```python
import numpy as np
```

```python
vazio = np.empty([3,2], dtype = int)
print(vazio)
print("-------")
zeros = np.zeros([4,3])
print(zeros)
print("-------")
um = np.ones([5,7])
print(um)
print("-------")
diagonal = np.eye(5)
print(diagonal)
```

```python
ale = np.random.random((5))
print(ale)
print("-------")
ale2 = np.random.randn((5))
print(ale2)
print("-------")
ale3 = (10*np.random.random((3,4)))
print(ale3)
```

```python
k = np.array([[17, 22, 43],[27, 25, 14],[15, 24, 32]])
print(k)
print(k[0][1])
print(k.shape)
```

```python
print(k.max())
print(k.min())
print(k.sum())
print(k.mean())
print(k.std())
```

```python
k1 = np.array([1, 4, 9, 16, 25, 36])
print(np.sqrt(k1))
print(np.exp(k1))
```

```python
m = np.array([1, 2, 3, 4, 5, 6])
print(m[1])
print("-------")
print(m[0:2])
print("-------")
print(m[1:])
print("-------")
print(m[-3:])
```

```python
l = np.array([[4, 5], [6, 1], [7, 4]])
print(l)
print("-------")
l_linha_1 = l[0, :]
print(l_linha_1)
print("-------")
l_linha_2 = l[1, :]
print(l_linha_2)
print("-------")
l_linha_3 = l[2, :]
print(l_linha_3)
print("-------")
l_coluna_1 = l[:, 0]
print(l_coluna_1)
print("-------")
l_coluna_2 = l[:, 1]
print(l_coluna_2)
```

```python
n = np.array([[1, 2], [3, 4]])
o = np.array([[1, 1], [1, 1]])
res1 = n+o
print(res1)
print("-------")
res2 = n*o
print(res2)
print("-------")
p = np.array([[1, 2], [3, 4], [5, 6]])
q = np.array([[2, 1]])
print(p+q)
```

```python
f = np.arange(15).reshape((3, 5))
print(f)
print("-------")
s = f.T
print(s)
```
