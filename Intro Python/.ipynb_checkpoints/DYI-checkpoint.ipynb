```python
import numpy as np

def amplitude(lst):
    max = lst.max()
    min = lst.min()
    amp = max - min
    print(amp)

amplitude(np.array([3,6,345,4,8,12,56,23]))
```

```python
def transposetxt(txt):
    for n in range(0, len(txt)):
       print(txt[n])

transposetxt("python yeah")
```

```python
def carga(peso):
    if peso <= 10:
        print("O valor sera 50 reais")
    elif peso > 10 and peso <= 20:
        print("O valor sera 80 reais")
    elif peso > 20:
        print("Transporte nao aceito")
        
pesostr = input("Informe o peso da carga: ")
pesoint = int(pesostr)
carga(pesoint)
```
