```python
import pandas as pd
from apyori import apriori
```

```python
dados = pd.read_csv('transacoes.txt', header = None)
dados
```

```python
#transformacao para o formato de lista, que e exigido pela biblioteca apyori
#6 e a quantidade de itens na base de dados
transacoes = []
for i in range(0,6):
    transacoes.append([str(dados.values[i,j]) for j in range(0,3)]) 
transacoes
```

```python
#execucao do algoritmo apriori para geracao das regras de associacao, definindo
#ps parametros de suporte e confianca
regras = apriori(transacoes, min_support = 0.5, min_confidence = 0.5, min_length=2)
```

```python
#criacao de nova variavel para armazenar somente as regras de associacao
resultados = list(regras)
print(resultados[0])
resultados
```

```python
#criacao de nova variavel, percorrendo a variavel anterior para uma melhor visualizacao
resultados2 = [list(x) for x in resultados]
resultados2
```

```python
#criacao de outra variavel para a visualizacao das regras ficar mais facil para o usuario
resultados3 = []
for j in range(0,7):
    resultados3.append([list(x) for x in resultados2[j][2]])
resultados3
```
