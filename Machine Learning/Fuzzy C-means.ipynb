```python
from sklearn import datasets
import numpy as np
from sklearn.metrics import confusion_matrix
import skfuzzy
```

```python
iris = datasets.load_iris()
```

```python
#Aplicacao do algoritmo definindo tres clusters (c = 3) e passando 
#a matriz transposta (iris.data.T). Os utros parametros sao obrigatorios e sao
#os default indicados na documentacao
r = skfuzzy.cmeans(data = iris.data.T, c = 3, m = 2, error = 0.005,
                  maxiter = 1000, init = None)
```

```python
previsoes_porcentagem = r[1]
```

```python
for x in range(150):
    print(previsoes_porcentagem[0][x], previsoes_porcentagem[1][x], previsoes_porcentagem[2][x])
```

```python
previsoes = previsoes_porcentagem.argmax(axis = 0)
resultados = confusion_matrix(iris.target, previsoes)
resultados
```
