```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import statsmodels.formula.api as sm
```

```python
base = pd.read_csv('mt_cars.csv')
base.shape
```

```python
base.head()
```

```python
x = base.iloc[:, 2].values
y = base.iloc[:, 0].values
correlacao = np.corrcoef(x, y)
correlacao
```

```python
#mudanca do x para formato de matriz
x = x.reshape(-1, 1)
```

```python
modelo = LinearRegression()
modelo.fit(x, y)
```

```python
modelo.intercept_
```

```python
modelo.coef_
```

```python
#score = R^2
modelo.score(x, y)
```

```python
previsoes = modelo.predict(x)
previsoes
```

```python
#criando modelo com biblioteca statsmodel
modelo_ajustado = sm.ols(formula = 'mpg ~ disp', data = base)
modelo_treinado = modelo_ajustado.fit()
modelo_treinado.summary()
```

```python
plt.scatter(x, y)
plt.plot(x, previsoes, color = 'red')
```

```python
modelo.predict([[200]])
```

```python
#criacao de novas variaveis x1 e y1 e novo modelo para comparacao com o anterior
#3 variaveis dependentes para prever mpg: cyl -> disp -> hp
x1 = base.iloc[:, 1:4].values
x1
```

```python
y1 = base.iloc[:, 0].values
modelo2 = LinearRegression()
modelo2.fit(x1, y1)
modelo2.score(x1, y1)
```

```python
modelo_ajustado2 = sm.ols(formula = 'mpg ~ cyl + disp + hp', data = base)
modelo_treinado2 = modelo_ajustado2.fit()
modelo_treinado2.summary()
```

```python
novo = np.array([4, 200, 100])
novo = novo.reshape(1, -1)
modelo2.predict(novo)
```
