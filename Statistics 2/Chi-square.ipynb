```python
import numpy as np
from scipy.stats import chi2_contingency
```

```python
novela = np.array([[19, 6], [43, 32]])
novela
```

```python
#Segundo valor e o pvalue
#Se valor de p e maior que 0,05, 
#nao ha evidencias de diferenca significativa (hipotese nula)
chi2_contingency(novela)
```

```python
novela2 = np.array([[22, 3], [43, 32]])
novela
```

```python
#Se valor de p menor que 0,05, pode-se rejeitar a hipotese nula em favor 
#da hipotese alternativa, que ha diferenca significativa
chi2_contingency(novela2)
```
