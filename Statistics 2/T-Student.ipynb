```python
from scipy.stats import t

#Média de salarios é R$75,00 por hora, com amostra de 9 funcionários
# e desvio padrão de 10.

#Graus de liberdade: 9-1 = 8
#stdev = 10
#Tabela T = 2.3
```

```python
#Probabilidade de selecionar um profissional e seu salario ser menor que 80 por hora.
t.cdf(2.3, 8)
```

```python
#Probabilidade de selecionar um profissional e seu salario ser maior que 80 por hora.
t.sf(2.3, 8)
```

```python
#Soma da execução dos dois códigos acima (lado esquerdo + lado direito)
t.cdf(2.3, 8) + t.sf(2.3, 8)
```
