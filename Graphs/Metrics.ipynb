```python
from igraph import Graph
from igraph import plot
import igraph
```

```python
grafo = igraph.load('Grafo.graphml')
print(grafo)
```

```python
plot(grafo, bbox = (0,0,600,600))
```

```python
grafo.degree(mode = 'all')
#grafo.degree(mode = 'in')
#grafo.degree(mode = 'out')
```

```python
grau = grafo.degree(mode = 'in')
print(grau)
```

```python
plot(grafo, vertex_size = grau, bbox = (0,0,600,600))
```

```python
#retorno di diametro do grafo (maior distancia entre vertices)
grafo.diameter(directed = True)
```

```python
#retorno dos vertices que possuem a maior distancia entre os pontos do grafo
grafo.get_diameter()
```

```python
#retorno dos vizinhos de cada vertice
grafo.neighborhood()
```

```python
#verificar se o grafo e isomorfico
grafo2 = grafo
grafo.isomorphic(grafo2)
```
